```python
import pandas as pd
import numpy as np

import joblib
from pathlib import Path
```

```python
project_root = Path().cwd().parent
```

```python
model = joblib.load(project_root/"models"/"churn_model.pkl")
```

```python
threshold = joblib.load(project_root/"models"/"threshold.pkl")
```

```python
def predict_churn(customer_data):
    
    probability = model.predict_proba(customer_data)[:, 1][0]
    
    prediction = int(probability >= threshold)
    
    return probability, prediction
```

```python
customer = pd.DataFrame([{
    "gender": "Male",
    "SeniorCitizen": 0,
    "Partner": "Yes",
    "Dependents": "No",
    "tenure": 5,
    "PhoneService": "Yes",
    "MultipleLines": "No",
    "InternetService": "Fiber optic",
    "OnlineSecurity": "No",
    "OnlineBackup": "No",
    "DeviceProtection": "No",
    "TechSupport": "No",
    "StreamingTV": "No",
    "StreamingMovies": "No",
    "Contract": "Month-to-month",
    "PaperlessBilling": "Yes",
    "PaymentMethod": "Electronic check",
    "MonthlyCharges": 80.0,
    "TotalCharges": 400.0
}])
```

```python
print(predict_churn(customer))
```
